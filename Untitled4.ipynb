```python
item_name = "Headphones"
item_price = 500
inventory_count = 10
wallet_balance = 1000

# 1. Check if the inventory is available for the item
if inventory_count > 5:
    # 2. Check if item_price is less than or equal to wallet_balance
    if item_price <= wallet_balance:
        wallet_balance -= item_price
        inventory_count -= 3
        print(f"Purchase successful! The new balance is ₹{wallet_balance}")
    else:
        print("Insufficient balance")
# 3. Else print the item is out of stock
else:
    print(f"The {item_name} is out of stock")
```
